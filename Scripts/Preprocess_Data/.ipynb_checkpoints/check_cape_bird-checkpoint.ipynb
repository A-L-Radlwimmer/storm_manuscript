# Storm Indices and Ice Fractions

```python
my_year = 2011
my_polynya = 'TNBP'    # RSP, MSP, TNBP
```

```python
import storm_indices as si

import numpy as np
import pandas as pd
import tqdm
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import seaborn as sns
import glob as gl
from datetime import datetime, date, time, timedelta
import os
import warnings
warnings.filterwarnings('ignore')
```

### Read in Temperature and Pressure Data from Scott Base

```python
if my_polynya == 'MSP':

    # SCOTT BASE
    t_column_name = 'Tair(C)'
    p_column_name = 'Pmsl(hPa)'
    station_name = 'ScottBase'
    
    TP = pd.read_csv('../../Data/ScottBase/ScottBase.csv')
    TP['DateTime'] = pd.to_datetime(TP['DateTime'],format='%Y-%m-%d %H:%M:%S')
    TP.index = TP['DateTime']
    del TP['DateTime']
    TP = TP.resample('1H').asfreq()
    TP = TP.interpolate(method='time', limit=2)
    
    TP['MeridionalWindSpeed(m/s)'] = TP['MeridionalWindSpeed(m/s)'].rolling('1D').mean()
    TP['off_coast_wind_speed_(m/s)'] = np.where(TP['MeridionalWindSpeed(m/s)'] > 0, 0, np.abs(TP['MeridionalWindSpeed(m/s)']))
    
    TP = TP[[t_column_name, p_column_name, 'off_coast_wind_speed_(m/s)']]
```

```python
if my_polynya in ['RSP','TNBP']:
    
    # AMRC AWS: VITO, MARBLEPOINT, MANUELA
    if my_polynya == 'RSP':
        station_name = 'Vito'
        TP = pd.read_csv('../../Data/Vito/Vito.csv')
    elif my_polynya == 'TNBP':
        station_name = 'Manuela'
        TP = pd.read_csv('../../Data/Manuela/Manuela.csv')
    t_column_name = 'temperature'
    p_column_name = 'pressure'
    
    TP['DateTime'] = pd.to_datetime(TP['DateTime'],format='%Y-%m-%d %H:%M:%S')
    TP.index = TP['DateTime']
    for i in ['DateTime', 'year', 'doy', 'month', 'day', 'hour', 'relative_humidity', '?']:
        del TP[i]
    TP = TP.resample('1H').asfreq()
    TP = TP.interpolate(method='time', limit=2)

    if my_polynya == 'RSP':
        TP['MeridionalWindSpeed(m/s)'] = np.cos(np.deg2rad(TP['wind_direction'])) * TP['wind_speed']
        TP['MeridionalWindSpeed(m/s)'] = TP['MeridionalWindSpeed(m/s)'].rolling('1D').mean()
        TP['off_coast_wind_speed_(m/s)'] = np.where(TP['MeridionalWindSpeed(m/s)'] > 0, 0, np.abs(TP['MeridionalWindSpeed(m/s)']))
    elif my_polynya == 'TNBP':
        TP['ZonalWindSpeed(m/s)'] = np.sin(np.deg2rad(TP['wind_direction'])) * TP['wind_speed']
        TP['ZonalWindSpeed(m/s)'] = TP['ZonalWindSpeed(m/s)'].rolling('1D').mean()
        TP['off_coast_wind_speed_(m/s)'] = np.where(TP['ZonalWindSpeed(m/s)'] > 0, 0, np.abs(TP['ZonalWindSpeed(m/s)']))
    
    TP = TP[[t_column_name, p_column_name, 'off_coast_wind_speed_(m/s)']]
```

```python
start_year = np.min(TP.index.year)
end_year = np.max(TP.index.year)
print(start_year, end_year)
TP
```

### Read in Daily Ice Fractions from ARTIST Analysis

```python
Ice_Fraction = pd.read_csv('Data/'+str(my_year)+'/IceFrac_'+str(my_polynya)+'_'+str(my_year)+'.csv', header=0)
Ice_Fraction['Hour'] = 12
Ice_Fraction['month_day_hour'] = Ice_Fraction['Month'].astype(str) + '-' + Ice_Fraction['Day'].astype(str) + '-' + Ice_Fraction['Hour'].astype(str)
unused = ['Year', 'Month', 'Day', 'Hour']
for i in unused:
    del Ice_Fraction[i]
Ice_Fraction.set_index('month_day_hour', inplace=True)
Ice_Fraction
```

### Compute High-Resolution Climatologies

```python
T_clim, P_clim = si.compute_climatologies_high_res(TP, t_column_name, p_column_name)
print(type(T_clim))
T_clim
```

```python
t_clim_df = T_clim.reset_index(name='T_clim')
p_clim_df = P_clim.reset_index(name='P_clim')
clim_df = pd.merge(t_clim_df, p_clim_df, on=['month', 'day', 'hour'], how='left')
clim_df['month_day_hour'] = clim_df['month'].astype(str) + '-' + clim_df['day'].astype(str) + '-' + clim_df['hour'].astype(str)
clim_df.index = clim_df['month_day_hour']
clim_df.drop(columns=['month', 'day', 'hour', 'month_day_hour'], inplace=True)
clim_df
```

### Compute Storm Index

```python
TP = si.compute_si(TP, t_column_name, p_column_name, 'SI')
TP
```

### Prepare Second Dataframe for Plotting

```python
TP_plot = TP.copy()
```

```python
TP_plot['month'] = TP_plot.index.month
TP_plot['day'] = TP_plot.index.day
TP_plot['hour'] = TP_plot.index.hour
TP_plot['month_day_hour'] = TP_plot['month'].astype(str) + '-' + TP_plot['day'].astype(str) + '-' + TP_plot['hour'].astype(str)
unused = ['month', 'day', 'hour']
for i in unused:
    del TP_plot[i]
TP_plot
```

```python
grouped = TP_plot.groupby(TP_plot.index.year)
years = [i for i in range(start_year, end_year+1)]
print(years)
dfs = {year: grouped.get_group(year) for year in years}
for i in years:
    dfs[i].set_index('month_day_hour', inplace=True)
    dfs[i].rename(columns={t_column_name: 'T'+str(i), p_column_name: 'P'+str(i), 'off_coast_wind_speed_(m/s)': 'W'+str(i), 'SI': 'SI'+str(i)}, inplace=True)
```

```python
TP_plot = dfs[start_year]
for i in years[1:]:
    TP_plot = TP_plot.merge(dfs[i], how='outer', on='month_day_hour')
```

```python
index_list = TP_plot.index.to_list()
TP_plot['month'] = [int(i.split('-')[0]) for i in index_list]
TP_plot['day'] = [int(i.split('-')[1]) for i in index_list]
TP_plot['hour'] = [int(i.split('-')[2]) for i in index_list]
TP_plot = TP_plot.sort_values(by=['month', 'day', 'hour'])
TP_plot = TP_plot.drop(columns=['month', 'day', 'hour'])
```

```python
TP_plot
```

### Plot

```python
# format x-axis
month_starts = [f'{m}-1-0' for m in range(1, 13)]
month_start_positions = [TP_plot.index.get_loc(month_start) for month_start in month_starts]
month_start_labels = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
```

```python
fig, ax = plt.subplots(4,1, figsize=(20,8), sharex=True, sharey=False, gridspec_kw={'height_ratios': [4,4,3,3]})
plt.subplots_adjust(hspace=0.)

T_mean = TP_plot[[f'T{year}' for year in years]].mean(axis=1)
T_std = TP_plot[[f'T{year}' for year in years]].std(axis=1)
P_mean = TP_plot[[f'P{year}' for year in years]].mean(axis=1)
P_std = TP_plot[[f'P{year}' for year in years]].std(axis=1)
ax[2].fill_between(TP_plot.index, T_mean - T_std, T_mean + T_std, color='gray', alpha=0.9, label='Temperature Variance', zorder=2)
ax[3].fill_between(TP_plot.index, P_mean - P_std, P_mean + P_std, color='gray', alpha=0.9, label='Pressure Variance', zorder=2)
ax[2].plot(clim_df.index, clim_df['T_clim'], color='black', linestyle=':')
ax[3].plot(clim_df.index, clim_df['P_clim'], color='black', linestyle=':')

ax[0].plot(TP_plot.index, TP_plot['SI'+str(my_year)], color='red')
ax0 = ax[0].twinx()
ax0.plot(TP_plot.index, TP_plot['W'+str(my_year)], color='blue', linewidth=0.5)
ax0.axhline(0, color='black', linewidth=0.5)
ax[1].plot(Ice_Fraction.index, Ice_Fraction['Ice Fraction'], color='black')
ax[2].plot(TP_plot.index, TP_plot['T'+str(my_year)], color='black', zorder=1)
ax[3].plot(TP_plot.index, TP_plot['P'+str(my_year)], color='black', zorder=1)

for j in [0,1,2,3]:
    ax[j].axvspan('1-1-0', '4-1-0', facecolor='0.9', zorder=0)
    ax[j].axvspan('9-30-23', '12-31-23', facecolor='0.9', zorder=0)

# formatting
ax[0].set_title(str(station_name)+' - '+str(my_year), fontsize=28)
for j in [0,1,2,3]:
    ax[j].tick_params(labelsize=14)
ax0.tick_params(labelsize=14)
ax[0].set_xticks(ticks=month_start_positions, labels=month_start_labels, rotation=45)
ax[0].set_xlim(('1-1-0', '12-31-23'))
ax[0].set_ylim((0, 880))
ax0.set_ylim((0, 42))
y0_rel_ax0 = (0 - ax[0].get_ylim()[0]) / (ax[0].get_ylim()[1] - ax[0].get_ylim()[0])
ax0_range = ax0.get_ylim()[1] - ax0.get_ylim()[0]
new_lower_ax0 = -y0_rel_ax0 * ax0_range
new_upper_ax0 = ax0_range + new_lower_ax0
ax0.set_ylim(new_lower_ax0, new_upper_ax0)
ax[1].set_ylim((0, 1))
ax[0].set_ylabel('Storm Index', fontsize=14)
ax[1].set_ylabel('Sea Ice\nFraction', fontsize=14)
ax[2].set_ylabel('T in °C', fontsize=14)
ax[3].set_ylabel('MSLP\nin hPa', fontsize=14)
if my_polynya == 'TNBP':
    ax0.set_ylabel('Westerlies in [m/s]', fontsize=14)
else:
    ax0.set_ylabel('Southerlies in [m/s]', fontsize=14)
ax[0].yaxis.label.set_color('red')
ax[0].tick_params(axis='y', colors='red')
ax0.yaxis.label.set_color('blue')
ax0.tick_params(axis='y', colors='blue')

plt.savefig('Figures/'+str(station_name)+str(my_year), bbox_inches='tight')
plt.show()
```
