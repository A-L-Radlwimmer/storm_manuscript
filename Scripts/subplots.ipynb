```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.image as mpimg
import os
import numpy as np
```

```python
fig = plt.figure(figsize=(16, 13))
gs = gridspec.GridSpec(2, 4, figure=fig, width_ratios=[1, 1, 1.4, 0.6], height_ratios=[3, 1], wspace=0., hspace=0.)
fig.subplots_adjust(hspace=-0.5)

ax1 = fig.add_subplot(gs[0,:2])
ax2 = fig.add_subplot(gs[0,2:])
ax3 = fig.add_subplot(gs[1,:3])
ax4 = fig.add_subplot(gs[1,3])
axes = [ax1, ax2, ax3, ax4]

image_dir = "../Figures"
list_of_subplots = ['storm_paper_ice_ages_2017_Fraser.png', 'storm_paper_ice_ages_2017.png', '2017_comparison.png', 'storm_paper_ice_ages_legend_vertical.png']
path_list = [os.path.join(image_dir, i) for i in list_of_subplots]

for ax, img_path in zip(axes, path_list):
    img = mpimg.imread(img_path)
    ax.imshow(img)
    ax.axis('off')
ax1.set_title(r'$\bf{(a)}$', fontsize=16, x=0.01, y=0.988, ha='left')
ax2.set_title(r'$\bf{(b)}$', fontsize=16, x=0.01, y=0.988, ha='left')
ax3.set_title(r'$\bf{(c)}$', fontsize=16, x=0.01, y=0.988, ha='left')

plt.savefig('../Figures/all_2017', dpi=300, bbox_inches='tight')
plt.show()
```
