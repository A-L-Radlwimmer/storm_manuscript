# Storm Indices

```python
import storm_indices as si

import numpy as np
import pandas as pd
import tqdm
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.patches as patches
import seaborn as sns
import shapefile as shp
import cartopy.crs as ccrs
import rasterio
from rasterio.plot import reshape_as_image
import glob as gl
from datetime import datetime, date, time, timedelta
import calendar
import os
import warnings
warnings.filterwarnings('ignore')
```

```python
#station_name = 'ScottBase'
#station_name = 'Vito'
#station_name = 'Margaret'
#station_name = 'Ferrell'
#station_name = 'MarblePoint'
#station_name = 'CapeBird'
#station_name = 'Manuela'
#station_name = 'Alessandra'
station_name = 'Silvia'
```

### Read McMurdo Ice Data

```python
path = '../Data/ice'
# create list of file names
flight_files = gl.glob(path + "/*.csv")
# create data frame
columns = ['Year', 'Month', 'Day', 'Ice Fraction']
content = []
for filename in tqdm.tqdm(flight_files, total=len(flight_files)):
    df = pd.read_csv(filename, names=columns, header=0, sep=',')
    content.append(df)
ICE = pd.concat(content)
ICE['DateTime'] = pd.to_datetime(ICE[['Year', 'Month', 'Day']])
ICE.set_index('DateTime', inplace=True)
ICE = ICE[['Ice Fraction']]

ICE
```

### Read in Temperature and Pressure Data

```python
if station_name == 'ScottBase':
    # SCOTT BASE
    t_column_name = 'Mean Temperature [Deg C]'
    p_column_name = 'Mean sea level pressure [Hpa]'
    TP = pd.read_csv('../Data/scottbase/scottbase.csv')
    TP['DateTime'] = pd.to_datetime(TP['DateTime'],format='%Y-%m-%d')
    TP.index = TP['DateTime']
    del TP['DateTime']
elif station_name in ['Alessandra', 'Silvia']:
    if station_name == 'Alessandra':
        TP = pd.read_csv('../Data/alessandra/alessandra.csv')
    elif station_name == 'Silvia':
        TP = pd.read_csv('../Data/silvia/silvia.csv')
    t_column_name = 'temperature_C'
    p_column_name = 'pressure_hPa'
    TP['DateTime'] = pd.to_datetime(TP['DateTime'],format='%Y-%m-%d')
    TP.index = TP['DateTime']
    del TP['DateTime']
else:
    # AMRC AWS: VITO, MARBLEPOINT, MANUELA, PHOENIX, FERRELL, WHITLOCK, POSSESSIONISLAND, CAPEBIRD, MARGARET, GILL
    if station_name == 'Vito':
        TP = pd.read_csv('../Data/vito/vito.csv')
    elif station_name == 'MarblePoint':
        TP = pd.read_csv('../Data/marblePoint/marblePoint.csv')
    elif station_name == 'Manuela':
        TP = pd.read_csv('../Data/manuela/manuela.csv')
    elif station_name == 'Ferrell':
        TP = pd.read_csv('../Data/ferrell/ferrell.csv')
    elif station_name == 'CapeBird':
        TP = pd.read_csv('../Data/capebird/capebird.csv')
    elif station_name == 'Margaret':
        TP = pd.read_csv('../Data/margaret/margaret.csv')
    t_column_name = 'temperature'
    p_column_name = 'pressure'
    TP['DateTime'] = pd.to_datetime(TP['DateTime'],format='%Y-%m-%d')
    TP.index = TP['DateTime']
    del TP['DateTime']
```

```python
TP = TP[[t_column_name, p_column_name]]
TP = TP[~TP.index.isna()]

TP
```

### Get Start and End Year

```python
start_year = int(np.min(TP.index.year))
end_year = int(np.max(TP.index.year))
print(start_year, end_year)
years = np.array([int(i) for i in range(start_year, end_year+1)], dtype=int)
```

### Compute Storm Index

```python
TP = si.compute_si(TP, t_column_name, p_column_name, 'SI')
TP
```

### Add Ice Fraction

```python
TP = TP.join(ICE, how='outer')
TP.loc['2019-01-01':'2024-12-31']
```

### Compute Climatologies

```python
SI_clim,_ = si.compute_climatologies(TP, 'SI', p_column_name)
si_clim_df = SI_clim.reset_index(name='SI_clim')
```

```python
T_clim, P_clim = si.compute_climatologies(TP, t_column_name, p_column_name)
t_clim_df = T_clim.reset_index(name='T_clim')
p_clim_df = P_clim.reset_index(name='P_clim')
clim_df = pd.merge(t_clim_df, p_clim_df, on=['month', 'day'], how='left')
clim_df = pd.merge(si_clim_df, clim_df, on=['month', 'day'], how='left')
clim_df['month_day'] = clim_df['month'].astype(str) + '-' + clim_df['day'].astype(str)
clim_df.index = clim_df['month_day']
clim_df.drop(columns=['month', 'day', 'month_day'], inplace=True)
clim_df.index = clim_df.index.to_series().apply(
    lambda x: '-'.join(f'{int(part):02d}' for part in x.split('-'))
)
clim_df
```

# Map

```python
stations = {
    "ScottBase":     (-77.85, 166.77),
    "MarblePoint":   (-77.439, 163.754),
    "CapeBird":      (-77.217, 166.439),
    "Margaret":      (-79.981, -165.099),
    "Vito":          (-78.408, 177.829),
    "Ferrell":       (-77.784, 170.816),
    "Manuela":       (-74.946, 163.687),
    "Alessandra":    (-73.58, 166.62),
    "Silvia":        (-73.05, 169.60),
}
station_colors = ('navy', 'royalblue', 'aqua', 'darkgreen', 'limegreen', 'lightgreen', 'crimson', 'gold', 'orange')
```

```python
south_polar_custom = ccrs.Stereographic(
    central_longitude=0,
    central_latitude=-90,
    true_scale_latitude=-71,
    globe=ccrs.Globe(ellipse='WGS84')
)

fig, ax = plt.subplots(figsize=(15, 8), subplot_kw={'projection': south_polar_custom})
plt.suptitle(r'$\bf{(a)}$ Weather station locations', fontsize=26, x=0.145, y=0.94, ha='left')

ax.set_extent([-1000000, 1000000, -2100000, -1010000], crs=south_polar_custom)

# Use MODIS image as base
with rasterio.open('../../../Worldview/snapshot-2023-11-09T00_00_00Z.tif') as src:
    img_data = src.read()
    bounds = src.bounds
img = np.moveaxis(img_data, 0, -1)
if img.dtype != np.uint8:
    img = np.clip(img, 0, 1)
    img = (img * 255).astype(np.uint8)
extent = [bounds.left, bounds.right, bounds.bottom, bounds.top]
ax.imshow(img, origin='upper', extent=extent, transform=south_polar_custom, zorder=0)

# Load shapefile
sf = shp.Reader('../Data/Map/add_coastline_high_res_polygon_v7_8.shp')

# Plot shapefile
for shape in tqdm.tqdm(sf.shapeRecords(), total=len(sf.shapeRecords())):
    for i in range(len(shape.shape.parts)):
        i_start = shape.shape.parts[i]
        i_end = shape.shape.parts[i+1] if i < len(shape.shape.parts)-1 else len(shape.shape.points)
        x = [pt[0] for pt in shape.shape.points[i_start:i_end]]
        y = [pt[1] for pt in shape.shape.points[i_start:i_end]]
        ax.plot(x, y, color='black', alpha=0.8, linewidth=0.75, transform=south_polar_custom)
    
# Plot locations
for (name, (lat, lon)), color in zip(stations.items(), station_colors):
    ax.plot(lon, lat, marker='o', color=color, markersize=8, transform=ccrs.PlateCarree())
    if name in ('Margaret'):
        ax.text(lon-1, lat, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    if name in ('Vito'):
        ax.text(lon+4, lat-0.1, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('Ferrell'):
        ax.text(lon+6, lat+0.1, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('ScottBase'):
        ax.text(lon+8, lat-0.2, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('MarblePoint'):
        ax.text(lon-1, lat, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('CapeBird'):
        ax.text(lon+5, lat+0.2, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('Manuela'):
        ax.text(lon+6, lat-0.2, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('Alessandra'):
        ax.text(lon+7, lat-0.5, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
    elif name in ('Silvia'):
        ax.text(lon+4, lat, name, fontsize=14, color=color, transform=ccrs.PlateCarree())
        
# Scale bar
scalebar_length_km = 100  # in km
scalebar_length_m = scalebar_length_km * 1000
x0, x1 = -950000, -850000 # Coordinates for lower left
y = -2040000              # corner (in map units)
ax.plot([x0, x1], [y, y], transform=south_polar_custom, color='k', linewidth=3, solid_capstyle='butt')
ax.text((x0 + x1)/2, y - 20000, f'{scalebar_length_km} km', transform=south_polar_custom, ha='center', va='top', fontsize=12)

# North arrow
ax.annotate('N', xy=(0.039, 0.11), xycoords='axes fraction', fontsize=16, fontweight='bold', ha='center', va='center')
ax.arrow(0.059, 0.15, 0, -0.06, transform=ax.transAxes, width=0.003, head_width=0.01, head_length=0.02, fc='k', ec='k')

plt.savefig('../Figures/Map_stations', bbox_inches='tight')
plt.show()
```
