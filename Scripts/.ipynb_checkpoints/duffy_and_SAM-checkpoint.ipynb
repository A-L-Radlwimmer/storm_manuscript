# Duffy et al., 2024, polynya data
#### from https://doi.org/10.6084/m9.figshare.24768654

```python
column_names = ('date', 'ROSarea_comb','ROSn_poly_comb','ROSarea_coastal','ROSn_poly_coastal','ROSarea_oceanic','ROSn_poly_oceanic')

polynyas = pd.read_csv('../Data/duffy/results/results/polynya_daily_summary_0p5.csv')
polynyas['date'] = pd.to_datetime(polynyas['date'], format='%Y-%m-%d')
polynyas = polynyas.set_index('date')
polynyas = polynyas.resample('MS').mean()
polynyas['duffy'] = polynyas['ROSarea_comb']
polynyas = polynyas[['duffy']]

polynyas
```

```python
test = polynyas[polynyas['duffy'].isna()==True]
test
```

```python
polynyas = polynyas.interpolate(method='time')
test = polynyas[polynyas['duffy'].isna()==True]
test
```

# SAM data
#### from https://www.cpc.ncep.noaa.gov/products/precip/CWlink/daily_ao_index/aao/aao.shtml

```python
SAM = pd.read_csv('../Data/atmospheric_modes/SAM_ncep.txt', names=['year', 'month', 'sam'], sep='\s+')
SAM['date'] = pd.to_datetime(dict(year=SAM['year'], month=SAM['month'], day=1))
SAM = SAM.set_index('date')[['sam']]
#SAM = SAM.loc['1979-01-01':'2022-12-01']    # Set to same time frame as Duffy's polynya dataset.
SAM
```

# Storm Index data

```python
station_name = 'ScottBase'
t_column_name = 'Mean Temperature [Deg C]'
p_column_name = 'Mean sea level pressure [Hpa]'
TP = pd.read_csv('../Data/scottbase/scottbase.csv')
TP['date'] = pd.to_datetime(TP['DateTime'],format='%Y-%m-%d')
TP.index = TP['date']
del TP['date']
TP = TP[[t_column_name, p_column_name]]
TP = TP[~TP.index.isna()]
SI = si.compute_si(TP, t_column_name, p_column_name, 'si')
SI = SI.resample('MS').mean()
SI = SI[['si']]
#SI = SI.loc['1979-01-01':'2022-12-01']    # Set to same time frame as Duffy's polynya dataset.

SI
```

# Combine

```python
df = SI.join(SAM, how='outer')
df = df.join(polynyas, how='outer')
# Normalize
df = (df - df.mean()) / df.std()
df
```

# Plot
### Pearson Correlation for Smoothed and Unsmoothed SI-SAM

```python
window_width = 3 * 12   # x*12 months
```

```python
df_test = df.copy()
```

```python
r = df_test['si'].corr(df_test['sam'])
print(r)
df_test['si_smoothed'] = df_test['si'].rolling(window=window_width, center=True).mean()
df_test['sam_smoothed'] = df_test['sam'].rolling(window=window_width, center=True).mean()
r = df_test['si_smoothed'].corr(df_test['sam_smoothed'])
print(r)
```

### Fourier Coherence between SI and SAM

```python
from scipy.signal import coherence

# Drop NaNs and ensure even spacing
df_test = df_test.dropna(how='any')
x = df_test['si'].interpolate().to_numpy()
y = df_test['sam'].interpolate().to_numpy()

# Compute coherence (fs = 12 because 12 samples per year)
f, Cxy = coherence(x, y, fs=12, nperseg=min(256, len(x)//2))

# Convert frequency to period in years
period = 1 / f

# Mask for 1 month to 20 year periods (20 years is max for length of time series)
mask = (period >= 1/12)

plt.figure(figsize=(12,6))
plt.plot(period[mask], Cxy[mask])
plt.xlabel('Period (years)', fontsize=16)
plt.ylabel('Coherence', fontsize=16)
plt.grid(True)
#plt.xscale('log')

plt.savefig('../Figures/fourier', bbox_inches='tight')
plt.show()
```

### Wavelet Coherence between SI and SAM

```python
x = df_test['si'].interpolate().to_numpy()
y = df_test['sam'].interpolate().to_numpy()
dt = 1/12  # years
t = df_test.index.to_pydatetime()
t_years = df_test.index.year + (df_test.index.month - 1) / 12

# Normalize
x_norm = (x - np.mean(x)) / np.std(x)
y_norm = (y - np.mean(y)) / np.std(y)

# Compute wavelet coherence correctly
WCT, aWCT, coi, freq, signif = wavelet.wct(x_norm, y_norm, dt, dj=1/12, s0=2*dt, significance_level=0.95, wavelet='morlet')

# Convert frequency to period (years)
period = 1 / freq

# Significance
if signif.ndim == 1:
    signif = np.tile(signif[:, None], WCT.shape[1])
sig_mask = WCT >= signif
```

```python
if signif.ndim == 1:
    signif = np.tile(signif[:, None], WCT.shape[1])
sig_mask = WCT >= signif

fig, ax = plt.subplots(figsize=(12, 6))

cf = ax.contourf(t_years, period, WCT, levels=100, cmap='jet')
ax.contour(t_years, period, sig_mask, levels=[0.5], colors='k', linewidths=1)
ax.fill_between(t_years, period.max(), coi, color='white', alpha=0.5)

ax.set_yscale('log')
ax.set_xlabel('Time (years)')
ax.set_ylabel('Period (years)')
ax.set_ylim(3/12, 20)
fig.colorbar(cf, ax=ax, label='Wavelet Coherence')

plt.savefig('../Figures/wavelet', bbox_inches='tight')
plt.show()
```
